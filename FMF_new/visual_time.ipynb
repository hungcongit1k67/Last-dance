# So sanh thoi gian chay theo kich ban (Kich ban 2)
Moi bo trong so (w1, w2, w3) co 1 doan code ve 2 bieu do:
1. **Thoi gian chay pha 1** (s)
2. **Tong thoi gian chay thuat toan** (s)

Chi lay so lieu cua **ACO-GA-A***, **FMF (khong lam muot)**, **WP-MSDF (khong lam muot)**.
Truc Ox la ten kich ban.

```python
import numpy as np
import matplotlib.pyplot as plt

# Ham tien ich: ve 2 bieu do (pha 1 va tong thoi gian) cho 3 thuat toan theo kich ban
def plot_time_two_charts(scenarios, data, weight_text, fname_prefix, logscale=False):
    """
    scenarios : list ten kich ban  -> truc Ox
    data      : dict {algo: {'pha1': [...], 'total': [...]}}
    weight_text: chuoi mo ta bo trong so, hien thi tren tieu de
    """
    algos  = ['ACO-GA-A*', 'FMF', 'WP-MSDF']
    colors = ['#1f77b4', '#ff7f0e', '#2ca02c']
    x = np.arange(len(scenarios))
    width = 0.25

    metrics = [
        ('pha1',  'Th\u1eddi gian ch\u1ea1y pha 1 (s)'),
        ('total', 'T\u1ed5ng th\u1eddi gian ch\u1ea1y thu\u1eadt to\u00e1n (s)'),
    ]

    for metric, ylabel in metrics:
        fig, ax = plt.subplots(figsize=(9, 6))
        for i, algo in enumerate(algos):
            ax.bar(x + (i - 1) * width, data[algo][metric], width,
                   label=algo, color=colors[i])
        ax.set_xlabel('K\u1ecbch b\u1ea3n', fontsize=13)
        ax.set_ylabel(ylabel, fontsize=13)
        ax.set_title(ylabel + '  (' + weight_text + ')', fontsize=13)
        ax.set_xticks(x)
        ax.set_xticklabels(scenarios)
        ax.legend(fontsize=12)
        if logscale:
            ax.set_yscale('log')   # bat log neu chenh lech qua lon (ACO-GA-A* >> FMF, WP-MSDF)
        plt.tight_layout()
        plt.savefig(fname_prefix + '_' + metric + '.png', dpi=150)
        plt.show()

scenarios = ['scenario4', 'scenario5', 'scenario6', 'scenario7']
```
